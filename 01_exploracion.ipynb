```python
import pandas as pd
import os

# Configuración de rutas
CARPETA_DATOS = 'datos'
ARCHIVOS = [
    'DATOSE_1.csv',
    'DATOSE_2.csv',
    'DATOSE_3.csv',
    'RIEGOS_1.CSV'
]

def explorar_y_convertir(carpeta, archivos):
    print("INICIANDO EXPLORACIÓN Y CONVERSIÓN DE DATOS")
    print("="*60)

    for nombre_archivo in archivos:
        ruta_completa = os.path.join(carpeta, nombre_archivo)
        
        # 1. Cargar el archivo según su extensión
        try:
            if nombre_archivo.lower().endswith(('.xls', '.xlsx')):
                # Lee Excel
                df = pd.read_excel(ruta_completa)
            elif nombre_archivo.lower().endswith('.csv'):
                # Lee CSV (intentamos detectar separadores comunes)
                try:
                    df = pd.read_csv(ruta_completa)
                except:
                    # Si falla, a veces los CSV en español usan punto y coma
                    df = pd.read_csv(ruta_completa, sep=';')
            else:
                print(f"[!] Formato no soportado: {nombre_archivo}")
                continue
                
            print(f"\n>>> ANALIZANDO: {nombre_archivo}")
            print("-" * 40)

            # 2. Exploración de Datos (EDA)
            
            # Dimensiones
            n_filas, n_cols = df.shape
            print(f"Dimensiones: {n_filas} filas x {n_cols} columnas")
            
            # Tipos de datos y Nombres de columnas
            print("\n--- Tipos de Columnas ---")
            print(df.dtypes)
            
            # Valores Nulos
            print("\n--- Valores Nulos (Missing Values) ---")
            nulos = df.isnull().sum()
            print(nulos[nulos > 0] if nulos.sum() > 0 else "No hay valores nulos.")
            
            # Estadísticas Descriptivas (Solo numéricas)
            print("\n--- Estadísticas Descriptivas ---")
            print(df.describe().T) # Transponemos para leer mejor
            
            # Muestra de datos
            print("\n--- Primeras 3 filas ---")
            print(df.head(3))

            # 3. Conversión a CSV
            # Generamos el nuevo nombre (mismo nombre pero extensión .csv)
            nombre_base = os.path.splitext(nombre_archivo)[0]
            nuevo_nombre = f"{nombre_base}.csv"
            ruta_salida = os.path.join(carpeta, nuevo_nombre)
            
            # Guardamos como CSV estándar (utf-8, separado por comas)
            df.to_csv(ruta_salida, index=False, encoding='utf-8')
            print(f"\n[OK] Archivo guardado/normalizado en: {ruta_salida}")
            print("="*60)

        except Exception as e:
            print(f"[ERROR] No se pudo procesar {nombre_archivo}: {e}")

if __name__ == "__main__":
    # Verificar que la carpeta existe
    if os.path.exists(CARPETA_DATOS):
        explorar_y_convertir(CARPETA_DATOS, ARCHIVOS)
    else:
        print(f"Error: La carpeta '{CARPETA_DATOS}' no existe.")
```
